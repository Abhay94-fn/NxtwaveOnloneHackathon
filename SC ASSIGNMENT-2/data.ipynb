1. To understand and implement the basic working of an artificial neuron using Python.

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def step_activation(x):
    return 1 if x >= 0 else 0
```

```python
def artificial_neuron(inputs, weights, bias):
    # Summing junction
    net_input = np.dot(inputs, weights) + bias
    # Activation
    return step_activation(net_input)
```

```python
# Testing the neuron with an AND logic gate simulation
X = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])
W = np.array([0.5, 0.5])
b = -0.7

print("Input\tOutput")
for x in X:
    print(f"{x}\t{artificial_neuron(x, W, b)}")
```

```python
class SingleLayerNN:
    def __init__(self, input_size, output_size):
        np.random.seed(42)
        self.weights = np.random.randn(input_size, output_size)
        self.bias = np.zeros((1, output_size))

    def forward(self, inputs):
        z = np.dot(inputs, self.weights) + self.bias
        # Sigmoid activation
        return 1 / (1 + np.exp(-z))
```

6. To apply Gradient Descent for updating the weights of a simple neural network during training.

```python
X = np.array([1, 2, 3, 4])
y = np.array([2, 4, 6, 8])

weight = 0.0 
learning_rate = 0.01
```

```python
for epoch in range(1, 51):
    # Forward pass
    y_pred = weight * X
    loss = np.mean((y_pred - y) ** 2)
    
    # Calculate gradient: derivative of MSE w.r.t weight
    gradient = -2 * np.mean(X * (y - y_pred))
    
    # Update weight
    weight = weight - learning_rate * gradient
    
    if epoch % 10 == 0:
        print(f"Epoch {epoch}: Weight = {weight:.4f}, Loss = {loss:.4f}")
```

7. To understand and implement the basic concept of Backpropagation for training a neural network.

```python
def sigmoid(x): return 1 / (1 + np.exp(-x))
def sigmoid_deriv(x): return x * (1 - x)
```

```python
# XOR Dataset
X = np.array([[0,0], [0,1], [1,0], [1,1]])
y = np.array([[0], [1], [1], [0]])

np.random.seed(1)
W1 = np.random.uniform(-1, 1, (2, 2)) # Input to Hidden
W2 = np.random.uniform(-1, 1, (2, 1)) # Hidden to Output
lr = 0.5
```

```python
for epoch in range(5000):
    # Forward Pass
    hidden_layer = sigmoid(np.dot(X, W1))
    output_layer = sigmoid(np.dot(hidden_layer, W2))
    
    # Backpropagation
    output_error = y - output_layer
    output_delta = output_error * sigmoid_deriv(output_layer)
    
    hidden_error = output_delta.dot(W2.T)
    hidden_delta = hidden_error * sigmoid_deriv(hidden_layer)
    
    # Update Weights
    W2 += hidden_layer.T.dot(output_delta) * lr
    W1 += X.T.dot(hidden_delta) * lr

print("Final Predictions:\n", np.round(output_layer, 2))
```

8. To train a simple neural network model on a dataset using Keras/TensorFlow.

```python
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
```

```python
data = load_breast_cancer()
X_train, X_test, y_train, y_test = train_test_split(data.data, data.target, test_size=0.2, random_state=42)
```

```python
# Scale features
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
# 2. Hidden Layers & Activation Functions
model = MLPClassifier(
    hidden_layer_sizes=(32, 16),
    activation='relu',
    solver='adam',
    random_state=42
)
```

```python
# 3. Training Integration
model.set_params(solver='adam', max_iter=50)
history = model.fit(X_train_scaled, y_train)
validation_accuracy = model.score(X_test_scaled, y_test)
print('Validation accuracy:', validation_accuracy)
```
